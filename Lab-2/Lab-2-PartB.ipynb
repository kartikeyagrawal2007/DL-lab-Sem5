```python
import numpy as np
from sklearn.datasets import make_moons
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD

# Same data as Part A
X, y = make_moons(n_samples=400, noise=0.20, random_state=1)
y = y.reshape(-1, 1).astype(float)
X = (X - X.mean(0)) / X.std(0)

def build():
    model = Sequential([
        Input(shape=(2,)),
        Dense(16, activation='relu'),
        Dense(16, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    model.compile(
        loss='binary_crossentropy',
        optimizer=SGD(learning_rate=0.5),
        metrics=['accuracy']
    )
    return model

# Batch gradient descent (full batch)
model = build()
model.fit(X, y, epochs=200, batch_size=len(X), verbose=0)
print("Batch GD accuracy:", model.evaluate(X, y, verbose=0)[1])

# Mini-batch SGD (fresh model so it starts from scratch)
model = build()
model.fit(X, y, epochs=200, batch_size=16, verbose=0)
print("SGD accuracy:", model.evaluate(X, y, verbose=0)[1])
```
